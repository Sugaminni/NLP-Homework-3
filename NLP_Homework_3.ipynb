```python
# Imports pandas
import pandas as pd

# Builds the document as token lists
documents = [
    "it is going to rain today".split(),
    "today i am not going outside".split(),
    "NLP is an interesting topic".split()
]

# Creates the vocabulary
vocab = sorted(set(word for doc in documents for word in doc))

# Maps each word to a row/column index
word_to_idx = {word: i for i, word in enumerate(vocab)}

# Empty co-occurrence matrix
size = len(vocab)
# Makes a 14x14(word * word) metrix full of zeros
matrix = [[0 for _ in range(size)] for _ in range(size)]

window = 2

# For first, second, and third documents
# Iterates through doc 1 to 3 and looks at every word in the sentence
for doc in documents:
    for i, target_word in enumerate(doc):
        target_idx = word_to_idx[target_word]

        # Boundary safe, to prevent "space issues"
        start = max(0, i - window)
        end = min(len(doc), i + window + 1)

        for j in range(start, end):   # Visits all words but the target word
            if i == j:
                continue

            context_word = doc[j]
            context_idx = word_to_idx[context_word]
            matrix[target_idx][context_idx] += 1  # For each context word, it shows where the word occurred near the window size

df = pd.DataFrame(matrix, index=vocab, columns=vocab) # Prints results on a table
print(df)
```

Cosine Similarity Function

```python
import math

# Defines a function that takes 2 vectors
def cosine_similarity(vec1, vec2):
  dot_product = sum(a * b for a, b in zip(vec1, vec2)) # Dot product for A · B (a1*b1 + a2*b2 + a3*b3 + ...)
  # Vector Length
  norm1 = math.sqrt(sum(a * a for a in vec1)) # This calculates ||A|| (Length of vector A)
  norm2 = math.sqrt(sum(b * b for b in vec2)) # This calculates ||B|| (Length of vector B)

  # Avoid division by zero
  if norm1 == 0 or norm2 == 0:
    return 0.0  # If vector is all zeros, its length is 0

  return dot_product / (norm1 * norm2)  #consine(A,B) = (A·B) / (||A|| ||B||)

# Finding the most similar words
best_pair = None  # Two most similar words
best_score = -1 # Their similarity value

# Nested Loops that compares every pair of words
for i in range(len(vocab)):
  for j in range(i + 1, len(vocab)):  # Avoids duplicate comparisons
    # Computes similarity
    score = cosine_similarity(matrix[i], matrix[j]) # Vector for word i and word j
    # Updates the best pair
    if score > best_score:
        best_score = score
        best_pair = (vocab[i], vocab[j]) # If a better similarity score is found, update the best result

# Prints results
print("Most similar words:", best_pair)
print("Cosine similarity:", best_score)
```
